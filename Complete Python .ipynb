```python
"""amstrong number""" # a number that is equal to the sum of its own digits, each raised to the power of the total number of digits
num=int(input("enter your number:"))
length=len(str(num))
total=0
temp=num
while num>0:
    digit= num%10
    total +=digit**length
    num= num//10
if total==temp:
    print(f"the number {temp} is amstrong number")
else:
    print(f"the number {temp} is not amstrong number")
```

```python
"Strong number"
num=int(input("enter a number:"))#A strong number is a number where the sum of the factorials of its individual digits equals the number itself
dup = num
total=0
while num>0:
    rem=num%10
    fact=1
    for i in range(1,rem+1):
        fact = fact*i
    total=total+fact
    num=num//10
if total == dup:
    print(f"{dup} is a strong number")
else:
    print("not strong")
```

```python
"""pallindrome""" #A palindrome number is a number that remains the same when its digits are reversed
num=int(input("enter a num:"))
dup=num
rev=0
place=10**(len(str(num))-1)
while num>0:
    rem=num%10
    rev=rev+(rem*place)
    num=num//10
    place=place//10
print("reverse:",rev)
if dup==rev:
    print(f"pallindrome number")
else:
    print("not pallindrome")
```

```python
# finding GCD 
def calculate_gcd(x,y):
    #choose least number
    least=min(x,y)
    for i in range(1,least+1):
        if x%i==0 and y%i ==0:
            gcd = i
    return gcd
x=12
y=18
calculate_gcd(x,y)
```

```python
# Step 1: Define a procedure to calculate the Greatest Common Divisor (GCD)
def calculate_gcd(num1, num2):
    while num2 != 0:
        remainder = num1 % num2
        num1 = num2
        num2 = remainder
    return num1


# Step 2: Define a procedure to calculate the LCM using the GCD result
def calculate_lcm(num1, num2):
    # Standard mathematical formula: (a * b) / GCD(a, b)
    gcd = calculate_gcd(num1, num2)
    lcm = abs(num1 * num2) // gcd
    return lcm


# Step 3: Main execution block (the sequential steps)
first_number = 12
second_number = 9


# Call the functions
final_lcm = calculate_lcm(first_number, second_number)
final_gcd = calculate_gcd(first_number, second_number)
# Output the final result
print(f"The LCM of {first_number} and {second_number} is: {final_lcm}")
print(f"The GCD of {first_number} and {second_number} is: {final_gcd}")
```

```python
import tkinter as tk

# Function to update the expression in the entry box
def press(num):
    global expression
    expression = expression + str(num)
    equation.set(expression)

# Function to evaluate the final expression
def equalpress():
    global expression
    try:
        # eval evaluates the string expression directly
        total = str(eval(expression))
        equation.set(total)
        expression = total # Keep the result for further operations
    except ZeroDivisionError:
        equation.set("Error: Div by 0")
        expression = ""
    except:
        equation.set(" Error ")
        expression = ""

# Function to clear the entry box
def clear():
    global expression
    expression = ""
    equation.set("")

# Main window setup
if __name__ == "__main__":
    root = tk.Tk()
    root.title("Simple Calculator")
    root.geometry("300x400")
    root.configure(bg="#202020") # Dark background

    expression = ""
    equation = tk.StringVar()

    # Display Screen
    display = tk.Entry(
        root, 
        textvariable=equation, 
        font=("Arial", 20), 
        bd=10, 
        insertwidth=4, 
        width=14, 
        borderwidth=0, 
        justify="right",
        bg="#333333",
        fg="white"
    )
    display.grid(columnspan=4, ipady=15, padx=10, pady=10)

    # Button configurations (Text, Row, Column)
    buttons = [
        ('7', 1, 0), ('8', 1, 1), ('9', 1, 2), ('/', 1, 3),
        ('4', 2, 0), ('5', 2, 1), ('6', 2, 2), ('*', 2, 3),
        ('1', 3, 0), ('2', 3, 1), ('3', 3, 2), ('-', 3, 3),
        ('C', 4, 0), ('0', 4, 1), ('=', 4, 2), ('+', 4, 3),
    ]

    # Create and place buttons dynamically using a loop
    for (text, row, col) in buttons:
        # Decide action based on button text
        if text == '=':
            action = equalpress
            bg_color = "#FF9500" # Orange for equal
        elif text == 'C':
            action = clear
            bg_color = "#A5A5A5" # Grey for clear
        elif text in ['/', '*', '-', '+']:
            action = lambda x=text: press(x)
            bg_color = "#FF9500" # Orange for math symbols
        else:
            action = lambda x=text: press(x)
            bg_color = "#3A3A3A" # Dark grey for numbers

        btn = tk.Button(
            root, 
            text=text, 
            font=("Arial", 14, "bold"),
            bg=bg_color, 
            fg="white", 
            activebackground="#555555",
            activeforeground="white",
            borderwidth=0,
            command=action
        )
        # Use grid system to place buttons evenly
        btn.grid(row=row, column=col, sticky="nsew", padx=5, pady=5)

    # Make rows and columns expand evenly when window resizing
    for i in range(4):
        root.columnconfigure(i, weight=1)
    for i in range(5):
        root.rowconfigure(i, weight=1)

    # Start the app application loop
    root.mainloop()
```

```python
#find factorial of a sequence using generator 
def factorial(num):
    if num>0:
        yield 1
        fact=1
        for i in range(1,num+1):
            fact*=i
            yield fact
    else:
        raise StopIteration
num=5
gobj=factorial(num)
print(list(gobj))
```

```python
#find fibonacci series of a sequence using generator 
def fib(num):
    f,s=0,1
    for i in range(num):
        t=f+s
        yield f
        f=s
        s=t
num=5
gobj=fib(num)
print(list(gobj))
        
```

## list

```python
# Using a loop — preserves original order
numbers = [4, 2, 5, 2, 1, 4, 3, 5]

result = []
seen = set()

for num in numbers:
    if num not in seen:
        result.append(num)
        seen.add(num)

print(result)
print(seen)
```

```python
# using set and then typecasting into list
numbers = [4, 2, 5, 2, 1, 4, 3, 5]
s=list(set(numbers))
print(numbers)
print(s)
```

```python
#Using dict.fromkeys() — concise and preserves order
numbers = [10, 20, 10, 30, 20, 40, 30]

unique_numbers = list(dict.fromkeys(numbers))

print(unique_numbers)
```

## string

```python
# using set - order may vary

s='12233345'
s1=set(s)
print(s)
print(s1)
```

```python
# using loop
text = "programming"

unique_text = ""

for char in text:
    if char not in unique_text:
        unique_text += char

print(unique_text)
```

```python
#Using dict.fromkeys() — concise and preserves order
text = "programming"

unique_text = "".join(dict.fromkeys(text))

print(unique_text)
```

## tuple

```python
numbers = (10, 20, 10, 30, 20, 40)

unique_numbers = tuple(dict.fromkeys(numbers))

print(unique_numbers)
```
